DATA DISTRIBUTION

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
values = np.random.uniform(low = -10.0, high = 10.0, size = 100000)
plt.hist(values, bins=50)
plt.show()
```

```python
from scipy.stats import norm

x = np.arange(1,5,1)
y = norm.pdf(x) # probability density function of the standard deviation normal distribution
print("x values: ", x)
print("y values: ", y)
plt.plot(x,y)
plt.show()

# x = 1 → y = 0.242
# x = 2 → y = 0.054
# x = 3 → y = 0.0044
# x = 4 → y = 0.00013

# 
```

```python
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

plt.plot(x, y)
plt.show()
```

```python
values = np.random.uniform(low=-10.0, high=10.0, size=10) 
print("Uniformly distributed values: ", values)
plt.hist(values, bins=10, edgecolor='black')
plt.title("Histogram of Uniformly Distributed Values")
plt.show()
```

```python
x = np.arange(-10.0, 10.0, 0.1)
# y = np.random.normal(loc=0.0, scale=1.0, size=len(x))

y = norm.pdf(x,scale=1,loc=0)
plt.plot(x, y)
plt.title("Normal/Gassian Distribution Approximation")
plt.show()
```
