```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
```

```python
score = [50, 60, 70, 75, 75, 75, 75, 85, 90, 95]
target = 75
# strict percentile considers only the scores strictly less than the target
'''
percentile = 3/10 * 100 = 30.0%
'''
# rank percentile considers the rank of the target score among all scores
'''
percentile_rank = (4+5+6+7)/4 = 5.5 => 5.5/10 * 100 = 55.0%
'''

percentile_of_target_strict = stats.percentileofscore(score, target, kind='strict')
percentile_of_target_rank = stats.percentileofscore(score, target, kind='rank')
print(f"Percentile of {target} (strict): {percentile_of_target_strict:.1f}th percentile")
print(f"Percentile of {target} (rank): {percentile_of_target_rank:.1f}th percentile")
```

```python
vals = np.random.normal(0,0.5,10000)
plt.hist(vals, bins=50)
plt.show()
```

```python
# find 50, 90, 20 percentiles
print("20th percentile",np.percentile(vals, 20))
print("50th percentile",np.percentile(vals, 50))
print("90th percentile",np.percentile(vals, 90))
print("-"*60)
percentiles = [20,50,90]
percentile_values  = np.percentile(vals, percentiles)
for p, v in zip(percentiles, percentile_values):
    print(f"{p}th percentile: {v}")
    
plt.axvline(percentile_values[0], color='r', linestyle='dashed', linewidth=1)
plt.axvline(percentile_values[1], color='g', linestyle='dashed', linewidth=1)
plt.axvline(percentile_values[2], color='b', linestyle='dashed', linewidth=1)
plt.show()
```
