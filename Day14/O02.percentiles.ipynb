- 9 students score above 85

```python
'''
what is a percentile?
A percentile is a measure used in statistics indicating the value below which a given
percentage of observations in a group of observations falls. 
For example, the 25th percentile is the value below which 25% of the observations may be found.
'''

def percentile(scores, target_score):
    less_count = sum(1 for s in scores if s < target_score)
    percentile = (less_count / len(scores)) * 100
    return percentile

test_scores = [min(84, 50 +i) for i in range(90)] + [85] + [min(99,86 + i) for i in range(9)]
# print("Test scores:", test_scores)

myscore= 85
result = percentile(test_scores, myscore)
print(f"Percentile of {myscore}: {result:.1f}th percentile")


from scipy import stats
scipy_result = stats.percentileofscore(test_scores, myscore,kind='strict')
print(f"Percentile of {myscore} using scipy: {scipy_result:.1f}th percentile")
```

```python
import numpy as np
np.random.seed(42)

class_marks = np.sort(np.random.randint(40, 100, size=100))
print("Class marks:", class_marks)

your_score = 87  
p25 = np.percentile(class_marks, 25)
p50 = np.percentile(class_marks, 50)
p75 = np.percentile(class_marks, 75)
p77 = np.percentile(class_marks, 77)
print(f"25th percentile: {p25}")
print(f"50th percentile (median): {p50}")
print(f"75th percentile: {p75}")
print(f"77th percentile: {p77}")
print(f"Your score: {your_score}")
your_percentile = np.percentile(class_marks, your_score)
print(f"Your score is at the {your_percentile}th percentile")
percentile_87 = np.mean(class_marks < your_score) * 100
print(f"Your score is at the {percentile_87}th percentile (calculated using mean method)")
```
