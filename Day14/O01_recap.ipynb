```python
# empirical verification of Bessel's correction via monte carlo simulation
chips = np.random.normal(size=100000,loc= 100.0, scale=15.0)
true_variance = np.var(chips)
sample_variance = np.var(chips, ddof=1)  # unbiased estimator of variance : ddof -> delta degrees of freedom
sample_size = 10
num_trials = 1000
biased_variance = []
unbiased_variance = []
for _ in range(num_trials):
    sample = np.random.choice(chips, size=sample_size, replace = False)
    biased_variance.append(np.var(sample,ddof=0)) # N
    unbiased_variance.append(np.var(sample,ddof=1)) # N-1
print("true variance:", true_variance)
print("sample_variance: ", sample_variance)
print(f"Unbiased Estimation (N-1) : {np.mean(unbiased_variance):.4f}")  
print(f"Biased Estimation (N)     : {np.mean(biased_variance):.4f}")
```

```python
import numpy as np
chips = np.random.normal(size = 100000, loc = 100.0, scale=15.0)
true_variance = np.var(chips)
sample_variance = np.var(chips, ddof=1)

def welford_algo(data):
    Mk = 0
    Sk = 0
    for k,x in enumerate(data, start=1):
        old_mean = Mk
        Mk = old_mean + ((x - old_mean) / k)
        Sk = Sk + (x - old_mean) * (x - Mk)
        
    biased_variance = Sk / len(data)
    unbiased_variance = Sk / (len(data) - 1)
    
    return biased_variance, unbiased_variance

sample_size = 10
num_trials = 1000

biased_variance = []
unbiased_variance = []

for _ in range(num_trials):
    sample = np.random.choice(chips, sample_size, replace=False)
    biased, unbiased = welford_algo(sample)
    biased_variance.append(biased)
    unbiased_variance.append(unbiased)
    
print("true variance:", true_variance)
print(f"Unbiased Estimation (N-1) : {np.mean(unbiased_variance):.4f}")
print(f"Biased Estimation (N)     : {np.mean(biased_variance):.4f}")
```
